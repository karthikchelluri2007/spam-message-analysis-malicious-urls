```python
from pathlib import Path

# Project root = parent of this notebook's stage folder (run the notebook from inside its own folder)
ROOT = Path.cwd().parent
RAW_CSV = ROOT / "dataset" / "dataset.csv"
INTER = ROOT / "dataset" / "intermediate"      # output of each stage becomes input of the next
RESULTS = ROOT / "results"
GRAPH_DIR = RESULTS / "graphs"
for p in (INTER, RESULTS, GRAPH_DIR):
    p.mkdir(parents=True, exist_ok=True)
```

```python
import pandas as pd

df_final = pd.read_csv(INTER / "stage4_clean.csv", keep_default_na=False)
print("Loaded cleaned data:", df_final.shape)
```

## Stage 5 — Data Aggregation & Representation
Summarize the engineered features, both overall and grouped by URL type.

```python
print("Overall summary:")
print("Total URLs:", len(df_final))
print("Unique domains:", df_final["domain"].nunique())
print("Unique TLDs:", df_final["tld"].nunique())
print("HTTPS URLs:", df_final["has_https"].sum(), f"({df_final['has_https'].mean()*100:.1f}%)")
print("URLs containing an IP address:", df_final["has_ip_address"].sum())
print("URLs using a shortener:", df_final["has_shortener"].sum())
print("URLs containing @:", df_final["has_at_symbol"].sum())

df_final.describe(include="all").T
```

```python
# Aggregation BY CLASS — this is the comparison that actually matters for this dataset
agg_by_type = df_final.groupby("type").agg(
    count=("url", "count"),
    avg_url_length=("url_length", "mean"),
    avg_keyword_count=("keyword_count", "mean"),
    pct_https=("has_https", "mean"),
    pct_ip_address=("has_ip_address", "mean"),
    pct_shortener=("has_shortener", "mean"),
    avg_special_chars=("special_char_count", "mean"),
).round(3)

agg_by_type["pct_https"] *= 100
agg_by_type["pct_ip_address"] *= 100
agg_by_type["pct_shortener"] *= 100

agg_by_type
```

```python
agg_by_type.to_csv(RESULTS / "aggregation_by_type.csv")
print("Saved:", RESULTS / "aggregation_by_type.csv")
```
