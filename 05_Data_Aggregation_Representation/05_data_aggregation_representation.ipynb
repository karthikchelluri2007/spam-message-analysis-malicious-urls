# Stage 5 — Aggregation & Representation

Summarize the cleaned URL set and compare feature averages by type.

```python
from pathlib import Path
import pandas as pd
import numpy as np
HERE = Path.cwd().resolve()
PROJECT_ROOT = next((p for p in (HERE, *HERE.parents) if (p / "dataset").is_dir()), HERE)
DATA_DIR = PROJECT_ROOT / "dataset"
INTERMEDIATE_DIR = DATA_DIR / "intermediate"
RESULTS_DIR = PROJECT_ROOT / "results"
INTERMEDIATE_DIR.mkdir(parents=True, exist_ok=True)
RESULTS_DIR.mkdir(parents=True, exist_ok=True)
input_path = INTERMEDIATE_DIR / "stage04_cleaned.csv"
if not input_path.is_file():
    raise FileNotFoundError("Run the earlier stage notebooks first; missing " + str(input_path))
df = pd.read_csv(input_path, encoding_errors="replace")
print("Loaded:", input_path, "| shape:", df.shape)
```

```python
print("Total URLs:", len(df))
print("Unique domains:", df["domain"].nunique())
print("Unique TLDs:", df["tld"].nunique())
print("HTTPS URLs:", df["has_https"].sum(), f"({df['has_https'].mean() * 100:.1f}%)")
print("URLs with IP address:", df["has_ip_address"].sum())
print("URLs using shortener:", df["has_shortener"].sum())
print("URLs containing @:", df["has_at_symbol"].sum())
```

```python
# Average of each feature for every URL type
cols = ["url_length", "keyword_count", "has_https", "has_ip_address",
        "has_shortener", "special_char_count"]

summary = df.groupby("type")[cols].mean()
summary["count"] = df["type"].value_counts()

# these three are 0/1 columns, so the mean x 100 = percentage
for c in ["has_https", "has_ip_address", "has_shortener"]:
    summary[c] = summary[c] * 100

summary.round(3)
```
